- typu kwadratowa, pierwiastek kwadratowy
- typu logarytm, odwrócony logarytm

- inne (np. piłokształtna).

Aby lepiej zobaczyć w jaki sposób działają różne przekodowania LUT, skonstruujemy funkcję, która jako argumenty pobierać będzie obrazek oryginalny oraz tablicę przekodowania, a następnie na wspólnym rysunku będzie wyświetlać: funkcję, obraz wejściowy oraz wynik przekodowania.

Przypomnienie składni tworzenia funkcji w pythonie:
```{python}
  def nazwa_funkcji(para, metry):
  	# cialo funkcji
```

  - Stwórz nową funkcję i nazwij ją LUT.
  - Funkcja powinna przyjmować dwa parametry: obraz oraz tablicę przekodowania.
  - W ciele funkcji wykonaj przekodowanie LUT, podobnie jak wcześniej w przykładzie.
  - Funkcja powinna wyświetlić wykres składający się z 3 umieszczonych obok siebie pól: wykres przekodowania, obraz oryginalny oraz obraz przekształcony.
    Każdy z wykresów powinien być podpisany.
    _(W razie problemów można przypomnieć sobie te zagadnienia z laboratorium wprowadzającego)_
  - Jeśli wykres przekodowania jest zbyt rozciągnięty, można go wyrównać, np. `ax2.set_aspect('equal')`.

````python
def LUT(image, table):

    plt.figure(figsize=(9,3))
    img_LUT = cv2.LUT(image, table)

    plt.subplot(1,3,1)
    plt.plot(table)
    plt.title("Tablica przekodowań")

    plt.subplot(1,3,2)
    plt.imshow(image)
    plt.title("Oryginalny obraz")

    plt.subplot(1,3,3)
    plt.imshow(img_LUT)
    plt.title("Przekształcony obraz")
````

Wywołaj przygotowaną funkcję dla każdego z przekształceń.
W wyniku powinno powstać 7 wykresów.

````python
przeksztalcenia = [
lut.log,
lut.pila,
lut.odwlog,
lut.wykladnicza,
lut.kwadratowa,
lut.odwrotna,
lut.pierwiastkowa
]

for i in przeksztalcenia:
    LUT(img,i)
````

## Operacja arytmetyczne

### Dodawanie

Wczytaj dwa obrazy _lena.bmp_ i _jet.bmp_ i wyświetl je.

````python
lena = cv2.imread("lena.bmp")
jet = cv2.imread("jet.bmp")

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(lena)
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(jet)
plt.axis("off")
plt.show()
````

Dodaj obrazy _lena_ i _jet_, wykorzystaj funkcję `cv2.add`.
Uzyskany wynik wyświetl.

````python
result = cv2.add(lena,jet)
plt.imshow(result)
````

Czy wynik sumowania jest satysfakcjonujący?
Co może niekorzystnie wpływać na rezultat operacji?
Spróbuj wykonać dodawanie ponownie wykorzystując typ uint16 (`jet.astype('uint16')`).
Wynikowy obraz należy przeskalować do zakresu 0-255, zamienić na typ uint8 i wyświetlić. _
Uwaga: operacja ta jest użyteczna w przypadku, gdy dane do wyświetlenia wykraczają poza zakres 0-255, w przeciwnym przypadku jej wykorzystanie może zniekształcić wyniki._

````python

lena16 = lena.astype('uint16')
jet16 = jet.astype('uint16')

result = cv2.add(lena16,jet16)
result = result / 2 # zeby nie przekraczac 255
result = result.astype('uint8')

plt.imshow(result)
plt.axis("off")

# Tym razem wynik jest zdecydowanie bardziej czytelny, wartosci nie przekraczaja tych 255
````

### Kombinacja liniowa

Do wykonywania operacji kombinacji liniowej służy funkcja `cv2.addWeighted`.
Zapoznaj się z dokumentacją tej funkcji i przetestuj kilka kombinacji liniowych obrazów _lena_ i _jet_.
Wyświetl wynik dowolnej z nich.

````python
alfa = 0.2
beta = 1.0 - alfa
dst = cv2.addWeighted(jet,alfa,lena,beta,0.0)

plt.imshow(dst)
plt.axis("off")
# Przetestowalem kilka kombinacji, uzywajac wagi w dodawaniu mówimy programowi ktory obraz ma byc "dominujacy" w sumie, w tym wypadku lena ktora ma wspolczynnik beta = 0.8
````

### Odejmowanie

Wykorzystując funkcję `cv2.subtract` odejmij obrazy _lena_ i _jet_.

````python
result = cv2.subtract(lena,jet)
plt.imshow(result)
plt.axis("off")
````

Czy wynik odejmowania jest satysfakcjonujący?
Co może niekorzystnie wpływać na rezultat operacji?
Często zamiast zwykłego odejmowania wykorzystuje się operację wartość bezwzględna z różnicy (pozwala to między innymi uniknąć pokazanych powyżej problemów).
Zamień typ obrazów _lena_ i _jet_ z uint8 na **int16**, odejmij je od siebie, a następnie wykorzystując funkcję `np.abs` wykonaj operację wartość bezwzględna z różnicy.
Wyświetl ten obraz.
Zastanów się, dlaczego ta zmiana poprawia wynik odejmowania?

````python
# Zmiana poprawila efekt koncowy bo pojawia sie podobny problem co przy dodawaniu, przepelnienie uint8
lena16 = lena.astype('int16')
jet16 = jet.astype('int16')
result = cv2.subtract(lena16,jet16)
result = np.abs(result)
result = result.astype('uint8')
plt.imshow(result)
plt.axis("off")
````

### Mnożenie

Mnożenie dwóch obrazów pozwala wykonać funkcja `cv2.multiply`.
Wykonaj mnożenie obrazów _lena_ i _jet_.
Czy wynik takiej operacji zawiera jakąś istotną informację?
Dlaczego?

````python
result = cv2.multiply(lena,jet)
plt.imshow(result)
plt.axis("off")

# Obraz wynikowy jest bialy bo mnozenie dwoch liczb 0-255 daje bardzo wysoki wynik co niesamowicie wykracza poza zakres
````

Mnożenie częściej wykorzystuje się jako:
  + mnożenie przez stałą $-$ co powoduje ogólne rozjaśnienie albo ściemnienie obrazu,
  + mnożenie przez maskę $-$ czyli obraz binarny.

Wczytaj maskę _kolo.bmp_ (https://raw.githubusercontent.com/vision-agh/poc_sw/master/02_Point/kolo.bmp).
Przemnóż wybrany obraz przez maskę i wyświetl wynik.
Mnożenie przez maskę można zrealizować za pomocą funkcji `cv2.multiply`, ale maskę należy należy najpierw przekształcić z zakresu 0-255 do 0-1, na przykład `(maska).astype('bool').astype('uint8')`.

````python
if  not os.path.exists("kolo.bmp"):
    !wget https://raw.githubusercontent.com/vision-agh/poc_sw/master/02_Point/kolo.bmp
maska = cv2.imread("kolo.bmp")
maska = maska.astype('bool').astype('uint8') # Zeby mnozenie nie wyskakiwalo poza skale
result = cv2.multiply(jet,maska)
plt.imshow(result)
plt.axis('off')

````

### Negatyw

Negatyw obrazu uzyskuje się za pomocą funkcji `cv2.bitwise_not(img)`
Negatyw obrazu można również uzyskać wykorzystując przekodowanie LUT.
Można w tym celu posłużyć się przygotowaną wcześniej tablicą `lut.odwrotna`.
Przetestuj działanie tych funkcji i wykonaj negatyw obrazu _lena_ dowolnym sposobem.

````python
neg = cv2.bitwise_not(lena)
neg_2 = cv2.LUT(lena,lut.odwrotna)

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(neg)
plt.axis("off")
plt.title('CV2 Negatyw')
plt.subplot(1,2,2)
plt.imshow(neg_2)
plt.axis("off")
plt.title('Negatyw metodą LUT')
plt.show()
````

## Operacje logiczne

Na poszczególnych punktach obrazu (najczęściej binarnego $-$ czyli składającego się z dwóch kolorów: czarnego i białego) można wykonywać operacje logiczne: NOT, AND, OR, XOR itp.
Wczytaj dwa obrazy: _kolo.bmp_ i _kwadrat.bmp_ (https://raw.githubusercontent.com/vision-agh/poc_sw/master/02_Point/kwadrat.bmp), następnie wyświetl je.

````python
if  not os.path.exists("kwadrat.bmp"):
    !wget https://raw.githubusercontent.com/vision-agh/poc_sw/master/02_Point/kwadrat.bmp

kolo = cv2.imread('kolo.bmp')
kwadrat = cv2.imread('kwadrat.bmp')

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(kolo)
plt.axis('off')
plt.subplot(1,2,2)
plt.imshow(kwadrat)
plt.axis('off')
````

Na wczytanych obrazach wykonaj wybrane operacje logiczne: NOT (operator `~`), AND (`&`), OR (`|`), XOR (`^`).
Operator `~` jest jednoargumentowy, wszystkie pozostałe są dwuargumentowe.
Alternatywnym sposobem jest wykorzystanie funkcji z biblioteki opencv: `cv2.bitwise_not`, `cv2.bitwise_and`, `cv2.bitwise_or`, `cv2.bitwise_xor`.
Wyświetl rezultaty.

````python
plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(~kolo)
plt.axis('off')
plt.title('NOT Koło')
plt.subplot(1,2,2)
plt.imshow(~kwadrat)
plt.axis('off')
plt.title('NOT Kwadrat')

plt.figure(figsize=(8,8))

plt.subplot(1,3,1)
plt.imshow(kolo & kwadrat)
plt.axis('off')
plt.title('Koło AND kwadrat')

plt.subplot(1,3,2)
plt.imshow(kwadrat | kolo)
plt.axis('off')
plt.title('Koło OR kwadrat')

plt.subplot(1,3,3)
plt.imshow(kwadrat^kolo)
plt.axis('off')
plt.title('Koło XOR kwadrat')
````
